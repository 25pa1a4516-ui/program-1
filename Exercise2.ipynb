```python
#2.check given year is leap year or not
year = int(input("enter year : "))
if(year%4==0 and year%100!=0):
  print(f"{year} is leap year")
elif(year%400==0):
  print(f"{year} is leap year")
else:
   print(f"{year} is not a leap year")

```

```python
# 3.match case
day  = int(input())
match day:
  case 1:
    print("monday")
  case 2:
    print("tuesday")
  case 3:
    print("wednesday")
  case 4:
    print("thursday")
  case 5:
    print("friday")
  case 6:
    print("saturday")

  case 7:
    print("sunday")
  case _:
     print("invalid")
```
