```python
#list operations
p1 = [34,78,90,"moon",23,12,7]
p2 = [67,89,10,"sun",33]
#concatenation
print(p1+p2)
#repetation
print(p1*2)
#indexing
print(p1[3])
print(p1[-5])
#slicing
print(p1[2:7])
print(p1[:8])
print(p1[5:])
print(p1[1:7:2])
print(p1[::2])
print(p1[:])
print(p1[::-1])
#membership operator
print(78 in p2)
print(6 not in p1)
if 78 in p1:
  print("true")
#iteration
for i in p1:
  print(i,end = " ")
```

```python
# builtin functions
b1 = [45,78,12,90,34,55]
b2 = [12,90,34,56]
#len()
print(len(b1))
#sum()
print(sum(b1))
#min()
print(min(b1))
#max()
print(max(b1))
#reversed()
print(list(reversed(b1)))
#sorted()
print(sorted(b1))
```

builtin methods

```python
#list methods(builtin methods)
#append(adding element to list)
b1.append(77)
print(b1)
```

```python
#extend(adding one list to other)
b1.extend(b2)
print(b1)
```

```python
#insert(adding element to specific index)
b2.insert(3,45)
print(b2)
```

```python
#pop(deleting last element)
b2.pop()
print(b2)
```

```python
#remove(deleting element)
b2.remove(90)
print(b2)
```

```python
#count
print(b2.count(34))
```

```python
#sort
b2.sort()
print(b2)
```

```python
#reverse
b2.reverse()
print(b2)
```

```python
#list comprehension
a = list(map(int,input().split()))
print(a)
```

```python
a = [i**2 for i in range(6)]
print(a)
```

```python
a = [int(i) for i in input().split()]
print(a)
```
