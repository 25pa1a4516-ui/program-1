DICTIONARIES

```python
# CREATING DICTIONARY
d = {"sno":116,"name":"rama","grade":7.5}
print(d)
```

```python
# USERDEFINED DICTIONARY
k = [i for i in input().split()]
v = [int(i) for i in input().split()]
d = {i:j for i,j in zip(k,v)}
print(d)
```

```python
# key present in dictionary or not
n = int(input())
d = {}
for i in range(n):
  k = input()
  v = input()
  d[k] = v
print(d)
key = input()
if key in d:
  print("yes")
else:
  print("no")
```

```python
# to add new key to the existing dictionary
d = {'id' : 16,'name':'moon','class':12}
d['branch'] = 'cse'
print(d)
```

```python
# print frequency or occurance of each element in a list in following format input{1 2 3 4 1 2 6 7 3 4} output {1:2,2:2,3:2,4:2,6:1,7:1}
l = list(map(int,input().split()))
d = {}
for i in l:
  key = i
  value = l.count(i)
  d[key] = value
print(d)
```
