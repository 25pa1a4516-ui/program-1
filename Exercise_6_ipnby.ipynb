TUPLES

```python
#Tuple Operations
t = (1,2,3,4,5)
t1 = (6,7,8,"MOON")
#concatenation
print(t+t1)
#repetation
print(t1*2)
#length
print(len(t))
#iteration
for i in t:
  print(i,end = " ")
#membership
if 5 in t:
  print("YES")
```

```python
# Tuple methods
print(t.count(1))
print(t1.index(8))
```

```python
#tuple builtin functions
print(len(t))
print(min(t))
print(max(t))
print(sum(t))
print(tuple(sorted(t)))
```

```python
# zip() function pairs up the items
t = ("n1","n2")
t1 = (1,2,3)
print(tuple(zip(t,t1)))
```

```python
t = (1,2,3,89,(56,78),[5,"MOON",23,78])
t[5][2] = "sun"
print(t)
```
