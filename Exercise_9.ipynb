File Handling Using Python

```python
# program on write mode
file = open("new.txt","w")
data = file.write("Hello world")
print("file created")
file.close()
```

```python
# program on read mode
with open("new.txt","r") as f:
  data = f.read()
  print(data)
```

```python
# program on append mode
with open("new.txt","a") as f:
  f.writelines(["\napple","\nname"])

```

```python
with open("new.txt","r") as f:
  print(f.readline())
```

```python
with open("new.txt","r") as f:
  print(f.readlines())
```

```python
with open("new.txt","r") as f:
  print(f.readline())
  print(f.tell())
  f.seek(1)
  print(f.tell())
  print(f.read())
```

```python
# copy contents from one file to another file
with open("new.txt","r") as f1:
  data = f1.read()
  print("done")
with open("f2.txt","w") as f2:
  f2.write(data)
  print("copied")
```

```python
# to print reverse
with open("new.txt","r") as f:
  data = f.readlines()
  for i in data:
    print(i[::-1])
```

```python
# write a code to copy contents and sort
with open("new.txt","r") as f1:
  data = f1.read().split()
  data_lower = [i.lower() for i in data]
  sort_data = sorted(data_lower)
with open("f2.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
  print("done")
```
