```python
# palindrome or not
def palindrome(a):
  i = len(a)-1
  j = 0
  while(i>=0):
    c = a[i]
    while(j<=i):
      if c==a[j]:
        return 1
      else:
        return 0
      j = j+1
    i = i-1
b = input()
if palindrome(b) == 1:
  print("Palindrome")
else:
  print("Not A Palindrome")
```

```python
# length of a string
def length(a):
  sum = 0
  for i in a:
    sum = sum+1
  return sum
b = input()
print(length(b))
```

```python
# function that returns multiple values
def multiple(*n):
  s = sum(n)
  l = len(n)
  a = s/l
  return s,l,a
s,l,a = multiple(1,3,6,8)
print(s,l,a)
```
