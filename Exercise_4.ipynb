```python
#string methods
a = str(input())
print(a)
b = a.split()
print(b)
c = '/'.join(b)
print(c)
print(a.upper())
print(a.lower())
print(a.strip())
print(len(a))
x = a.replace("hello","good")
print(x)
print(a.find("o"))
print(a.count("l"))
```

```python
# write a program that counts no of vowels in string
a = str(input())
c = 0
for i in a:
  if (i=='a' or i=='e' or i=='i' or i=='o' or i=='u' or i=='A' or i=='E' or i=='I' or i=='O' or i=='U' ):
    c = c+1
print(c)
```

```python
# program that checks substring is present in the given string
a = str(input())
b = str(input())
if b in a:
  print(a.find(b))
else:
  print("NO")
```

```python
#reverse a number using functions
def reverse(a):
  s = 0
  while(a!=0):
    r = a%10
    s = s*10+r
    a = a//10
  print(s)
b = int(input())
reverse(b)
```
